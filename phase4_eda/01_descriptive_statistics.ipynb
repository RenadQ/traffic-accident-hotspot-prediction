```python
import pandas as pd
import numpy as np

file_path = 'transformed_data_train_V88.xlsx'
output_file_path = 'descriptive_stats_summary.xlsx'

try:
    df = pd.read_excel(file_path)
    print(f"Dataset loaded successfully. Original shape: {df.shape}")
except FileNotFoundError:
    print(f"Error: The file '{file_path}' was not found. Please check the file name and path.")
    # Exit the script or notebook cell if the file cannot be loaded
    # raise 
    
# Identify Numerical Columns for Analysis
# We focus on continuous numerical measures (weather, distance, coordinates, Severity)
numeric_cols = [
    'Severity', 'Start_Lat', 'Start_Lng', 'End_Lat', 'End_Lng', 
    'Distance(mi)', 'Temperature(F)', 'Wind_Chill(F)', 
    'Humidity(%)', 'Pressure(in)', 'Visibility(mi)', 'Wind_Speed(mph)',
]


cols_to_analyze = [col for col in numeric_cols if col in df.columns]

# Calculate Descriptive Statistics
# We will create a dictionary to store the results, where each key is a statistic.
stats_dict = {}

# Calculate Mean, Median, and Variance 
stats_dict['Mean'] = df[cols_to_analyze].mean()
stats_dict['Median'] = df[cols_to_analyze].median()
stats_dict['Variance'] = df[cols_to_analyze].var()

# Calculate Mode
stats_dict['Mode'] = df[cols_to_analyze].mode().iloc[0] 

# Combine results into a clean DataFrame for reporting
summary_df = pd.DataFrame(stats_dict).T

#I want to add minimum, maximum, and quantity limits to get more comprehensive reports.
summary_df.loc['Min'] = df[cols_to_analyze].min()
summary_df.loc['Max'] = df[cols_to_analyze].max()
summary_df.loc['Count'] = df[cols_to_analyze].count()

print("\nDescriptive Statistics Summary (First 5 columns):\n")
print(summary_df.head().to_markdown(numalign="left", stralign="left"))


# Save the Summary to a NEW Excel File, This is a safe way to keep the original training file unchanged.
try:
    summary_df.to_excel(output_file_path, index=True) 
    print(f"\nSuccessfully saved descriptive statistics to: {output_file_path}")
except Exception as e:
    print(f"\nError during file saving: {e}")


# Alternative: Code for Direct Modification
# Note: This is left as a comment block, we will use it when needed.
"""
# CODE TO OVERWRITE THE ORIGINAL FILE (transformed_data_train_V88.xlsx)
# WARNING: This replaces the data in the original file with the summary table only.
# It's highly recommended to use the separate output file method above.
# summary_df.to_excel(file_path, index=True) 
"""
```
